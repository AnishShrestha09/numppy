```python
# A warehouse has 4 storage sections and 5 products.

# Initially, every product has a stock level of 100 units.

# Tasks
# Create a 4 x 5 array where every value is 100.
# Increase the stock of the first two sections by 20 units
# Reduce stock of the last two products by 10%.
# Display only the middle two sections.
# Display only the first three products.
# Display the bottom-right 2 x 2 portion of the matrix.
```

```python
# Create a 4 x 5 array where every value is 100.
import numpy as np
product = np.full((4,5),100)
print(product)

# Increase the stock of the first two sections by 20 units
product[:2] += 20
print(product)


# Reduce stock of the last two products by 10%.
product[:,-2:] = 0.9 * product[:,-2:]


# Display only the middle two sections.
print(product[1:3])

# Display only the first three products.
print(product[:,:3])

# Display the bottom-right 2 x 2 portion of the matrix.
print(product[-2:, -2:])

```

```python
# A company has salary data for 4 employees:

# salary = np.array([
# [30000, 5000],
# [40000, 6000],
# [50000, 8000],
# [60000, 10000]

# })

# Column 1 = Basic Salary
# Column 2 = Allowance

# Tasks
# Display the complete information of the 2nd employee.
# Display all basic salaries.
# Display all allowances.
# Increase all allowances by ₹1,000.
# Increase the basic salary of the last two employees # by 10%.
# Display the final salary matrix.

```

```python
salary = np.array([
[30000, 5000],
[40000, 6000],
[50000, 8000],
[60000, 10000]

])
```

```python
# Tasks
# Display the complete information of the 2nd employee.
print(salary[1])

# Display all basic salaries.
print(salary[:,0])

# Display all allowances.
print(salary[:,1])

# Increase all allowances by ₹1,000.
salary[:,1] += 1000

# Increase the basic salary of the last two employees # by 10%.
salary[-2:,0] = 1.1 *salary[-2:,0]

# Display the final salary matrix.
print(salary)

```

```python
# A startup expects its monthly users to grow from 10,000 to 100,000 over 12 months.

# Tasks
# Generate monthly user targets using np. linspace()
# Create month numbers using np. arange().
# Display users expected in Month 1 and Month 12.
# Display users expected from Months 4 - 8.
# Increase all targets by another 15%.
# Display the updated targets for the last 4 months.
```

```python
# Generate monthly user targets using np. linspace()
monthlyusers = np.linspace(10000, 100000,12)
print(monthlyusers)

# Create month numbers using np. arange().
months = np.arange(1,13)
print(months)

# Display users expected in Month 1 and Month 12.
print(monthlyusers[0])
print(monthlyusers[11])


# Display users expected from Months 4 - 8.
print(monthlyusers[3:8])


# Increase all targets by another 15%.
monthlyusers = monthlyusers * 1.15

# Display the updated targets for the last 4 months.
print(monthlyusers[-4:])
```
