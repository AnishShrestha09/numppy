```python
# A company records leads generated by four campaigns across five channels:
import numpy as np

leads = np.array([
[12000, 15000, 11000, 18000, 21000],
[14000, 13000, 17000, 19000, 22000],
[10000, 16000, 15000, 20000, 23000],
[18000, 17000, 19000, 21000, 25000]

])

# Find total leads generated by each campaign.
total_leads = np.sum(leads,axis=1)
print(total_leads)

# Find total leads generated by each channel.
total_leads1 = np.sum(leads,axis=0)
print(total_leads1)

# Find the campaign with maximum leads.
a= np.argmax(total_leads)
print(leads[a])


# Find the channel with maximum leads.
a= np.argmax(total_leads1)
print(a)


```

```python
# Find average leads per campaign.
average = np.average(leads,axis=1)
print(average)

# Find values above overall average.
overall_a = np.average(leads)
print(overall_a)
leads[leads>overall_a]

# Increase values below 15,000 by 10%.
increased_values = np.where(leads <15000,leads * 1.1,leads)
print(increased_values)

# Categorize values above median as "High" and others as "Low".
category = np.where(leads <np.median(leads),"high","low")
print(category)


# Extract every second channel.
leads[:,1::2]
```

```python
np.median(leads)
```

```python
import numpy as np
orders = np.array([
[101, 3, 1200, 4.5],
[102, 5, 800, 3.8],
[103, 2, 2500, 4.8],
[104, 8, 450, 4.1],
[105, 4, 1800, 3.5],
[106, 6, 950, 4.7],
[107,3,2200, 4.9],
[108, 10, 600, 3.2],
[109, 7, 1500, 4.6],
[110, 2, 3000, 4.0],
[111, 5, 750, 3.9],
[112, 9, 1250, 4.8]
])
# Order_ID
# Quantity
# Price
# Rating

```

```python
# Inspect dimensions, shape and size.
print(orders.ndim)

# Extract the first 5 orders.
print(orders[:5])

# Extract the last 3 orders.
print(orders[-3:])

# Extract every second order.
print()
print(orders[1::2])

# Calculate revenue for every order.
revenue = orders[:,1] *orders[:,2]
print(revenue)
```

```python
# Calculate total and average revenue.
np.sum(revenue)

# Find the order with highest revenue.

orders[np.argmax(revenue)]

# Calculate median revenue.
print(np.median(revenue))
```

```python
# Order_ID
# Quantity
# Price
# Rating

# Categorize orders:
# Revenue >= 15,000 High
# Revenue >= 7,500 + Medium
# Otherwise + Low

# Find orders where quantity >= 5 AND rating >= 4.5.
conditions= ( (orders[:,1]>=5) & (orders[:,-1]>=4.5))
print(orders[conditions])

# Find orders where price is above median OR rating is below 4.
condtion = ((orders[:,2] >np.mean(orders[:,2])) | ( orders[:,-1]<4))
orders[condtion]

# Give 10% discount to products whose price is above ₹1,000.

discountamount = np.where(orders[:,2]>=1000,0.1 *orders[:,2],0)
print(discountamount)
```

```python
# Round discounted prices to 2 decimals.
np.round(discountamount,2)

# Calculate standard deviation of price.
print(np.std(orders[:,2]))

# Find 25th and 75th percentile of price.
print(np.percentile(orders[:,2],75))

# Find prices above the 75th percentile.
orders[orders[:,2]>np.percentile(orders[:,2],75)]
```

```python
# percentage of orders with rating >= 3

total_orders = len(orders)
ratinggt3 = len(orders[:,-1]>=3)
print(ratinggt3)
percentage = ratinggt3/total_orders * 100
print(percentage)
```
