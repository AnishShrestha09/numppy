```python
# An online store has the following product prices:

# prices = np. array([
# 499.50, 799.90, 1299.75,
# 2499.50, 3999.90, 599.25,
# 899.80, 1599.40
# ])
# Perform:

# Round prices to the nearest whole number.
# Find products costing more than ₹1,000.
# Calculate a 15% discount only on products costing more than ₹1,000.
# Calculate the average discounted price of those products.
# Find the most expensive original product.
# Extract the last 4 products.
# Calculate the square root of the final prices and round them to 2 decimal places.
```

```python
import numpy as np
```

```python
prices = np. array([
499.50, 799.90, 1299.75,
2499.50, 3999.90, 599.25,
899.80, 1599.40
])
```

```python
# Round prices to the nearest whole number.
roundprice = np.around(prices)
print(roundprice)
```

```python
# Find products costing more than ₹1,000.

prices[prices > 1000]
```

```python
# Calculate a 15% discount only on products costing more than ₹1,000.
prices[prices > 1000] = 85 * prices[prices > 1000]
```

```python
prices
```

```python
# Calculate the average discounted price of those products.
np.average( 0.85 * prices[prices > 1000] )
```

```python
# Find the most expensive original product.
np.max(prices)
```

```python
# Extract the last 4 products.
prices[-4:]
```

```python
# Calculate the square root of the final prices and round them to 2 decimal places.
np.round(np.sqrt(prices),2)
```
