```python
campaigns = np.array([
[101, 10000, 800, 80, 5000, 7500],
[102, 15000, 1200, 150, 7000, 12000],
[103, 8000, 400, 30, 3000, 4200],
[104, 20000, 1800, 250, 9000, 18000],
[105, 12000, 600, 90, 6000, 8000],
[106, 18000, 1500, 220, 8000, 16000],
[107, 9000, 300, 20, 2500, 3000],
[108, 25000, 2200, 300, 11000, 24000]
])
```

```python
# Campaign_ID | Impressions | Clicks | Conversions | Cost | Revenue

# CTR = Clicks / Impressions * 100

# Conversion Rate = Conversions / Clicks * 100

# ROI = (Revenue - Cost) / Cost * 100
```

```python


print(" Find highest-revenue campaign.")
index = np.argmax(campaigns[:,-1])
print(campaigns[index])


print()


print("Find highest-ROI campaign.")
revenue = (campaigns[:,-1] - campaigns[:,-2]) / campaigns[:,-1] * 100
# print(revenue)
highest = np.argmax(revenue)
print(campaigns[highest])
print()
# Calculate average CTR.
print("Calculate average CTR.")
CTR = campaigns[:,2] / campaigns[:,1] * 100
print(np.average(CTR))
print()

# Calculate average ROI.
print(np.average(revenue))


# Find campaigns with above-average CTR and ROI.
```
