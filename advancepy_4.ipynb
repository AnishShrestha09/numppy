```python
# Categorize each student as:
# - High attendance: 85% or above
# - Moderate attendance: 70% to below 85%
# - Low attendance: below 70%
# - Unknown: attendance is missing
```

```python
attendacecat = np.where(attendance >= 85,"high",np.where(attendance >=70,"moderate",np.where(attendance<70,"low","missing")))
attendacecat
```

```python
conditions = [
    np.isnan(attendance),
    attendance > 85,
    attendance > 70
]

choices = [
    "unknown",
    "high",
    "moderate"
]

attendancecat = np.select(
    conditions,
    choices,
    default="low"
)

print(attendancecat)
```

```python
# Categorize students into three study bands: less than 2 hours, 2 4 hours inclusive, and more than 4 hours.

category = [
    study_hours > 4,
    (study_hours < 4)& (study_hours > 2)

]

choices = [
    "highlearning",
    "mediumlearning"
]


cat = np.select(
    category,
    choices,
    default="low"
)

print(cat)
```
