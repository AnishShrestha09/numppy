```python
import numpy as np

# A factory collects temperatures from 20 sensor readings:

temperature = np. array ([
28.5, 31.2, 29.8, 35.6, 38.1,
40.2, 27.9, 30.5, 42.3, 39.8,
36.4, 33.2, 29.1, 41.7, 44.2,
31.8, 34.5, 37.6, 26.9, 43.1
])
```

```python
# Calculate mean, median, minimum and maximum.
print(np.mean(temperature))
print(np.median(temperature))
print(np.max(temperature))

# Calculate variance and standard deviation.
print(np.square(np.std(temperature))) #variance
print(np.std(temperature))


# Find all temperatures more than one standard deviation above the mean.
print(temperature[temperature >np.std(temperature) ])

# Calculate the absolute difference between every temperature and the mean.
print(np.abs(temperature - np.mean(temperature)))

# Find the five readings having the largest deviation from the mean.
a = np.sort(temperature)
print("a")
print(a[-5:])

# Round all temperatures to the nearest whole number.
print(np.round(temperature))
```

```python
Categorize temperatures:
>= 40 "Critical"
>= 35 - "Warning"
Otherwise - "Normal"





```

```python
# Find the number and percentage of critical readings.
np.where(temperature>=40,"critical",np.where(temperature >=35,"warning","normal"))

critical = len(temperature[temperature>=40 ])
total = len(temperature)

critical_per = critical/total * 100
print(critical,critical_per)
```

```python
# Reshape the readings into a 4 × 5 array.
a= temperature.reshape(4,5)
print(a)
```

```python
# Find the average temperature for each row.
np.average(a,axis= 1)
```
