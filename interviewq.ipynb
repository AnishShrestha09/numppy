```python
# Ignore spaces, punctuation, and differences between uppercase and lowercase letters.
Input = "A man, a plan, a canal: Panama"


def cleaned(string1):
  cleaned = ""

  for c in string1.lower():
    if c.isalnum():
      cleaned+=c
      # print(c)

  return cleaned


a = cleaned(Input)
b = cleaned(Input[::-1])

if a == b:
  print("palindrome")
```

```python
# Problem statement: Two strings are anagrams
# if they contain the same characters with the same frequencies, regardless of order.
# Ignore spaces and case.
```

```python
s1 = "listen"
s2 = "silet"

frequency = {}
frequency2 = {}


for s in s1 :
  if s not in frequency:

    frequency[s] = 1
  else:
    frequency[s] += 1


for s in s2 :
  if s not in frequency2:
    frequency2[s] = 1
  else:
    frequency2[s] += 1

if (frequency == frequency2):
  print("same")
else:
  print("not same")

```

```python
# Problem statement: Given a string, find the first character that appears exactly once.
# Return -1 if every character repeats.
```

```python
string = "aaabd"
frequency = {}

for s in string :
  if s not in frequency:

    frequency[s] = 1
  else:
    frequency[s] += 1
# print(frequency)


for key in frequency:
  if(frequency[key]==1):
      print(key , frequency[key])
      break


```

```python
# Problem statement:
# Given two arrays sorted in ascending order,
# merge them into one sorted array without sorting the combined result afterward.

# Input:


l1 = [1, 4, 7, 9]
l2 = [2, 3, 8, 10]
l3 = []

i = 0
j = 0

while i < len(l1) and j < len(l2):
    if l1[i] < l2[j]:
        l3.append(l1[i])
        i = i + 1
    else:
        l3.append(l2[j])
        j = j + 1

l3.extend(l1[i:])
l3.extend(l2[j:])

print(l3)
```
