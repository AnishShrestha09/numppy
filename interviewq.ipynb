```python
# Find Two Numbers Whose Sum Equals a Target

# Problem statement: Given an array and a target value,
# return the indices of two distinct elements whose sum equals the target.
# Assume exactly one valid pair exists.
```

```python
arr = [5,6,7,8,9]
target = 11


for i in range(len(arr)-1):
  for j in range(1,len(arr)):
    sum =arr[i] + arr[j]
    if (sum==target):
      print(i, j)
```

```python
# Ignore spaces, punctuation, and differences between uppercase and lowercase letters.
Input = "A man, a plan, a canal: Panama"


def cleaned(string1):
  cleaned = ""

  for c in string1.lower():
    if c.isalnum():
      cleaned+=c
      # print(c)

  return cleaned


a = cleaned(Input)
b = cleaned(Input[::-1])

if a == b:
  print("palindrome")
```
