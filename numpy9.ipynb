```python
# A business records revenue for 12 months:

# revenue = np.array([
# 120000, 135000, 128000, 145000,
# 150000, 138000, 160000, 172000,
# 155000, 148000, 180000, 195000
# ])
# Find:

# Average monthly revenue
# Highest revenue
# Lowest revenue
# Month index with highest revenue
# Month index with lowest revenue
# Total annual revenue
# Difference between highest and lowest revenue
```

```python
import numpy as np
revenue = np.array([
120000, 135000, 128000, 145000,
150000, 138000, 160000, 172000,
155000, 148000, 180000, 195000
])
```

```python
# Average monthly revenue
np.average(revenue)

# Highest revenue
print(np.max(revenue))

# Lowest revenue
print(np.min(revenue))


# Month index with highest revenue
print(np.argmax(revenue))


# Month index with lowest revenue
print(np.argmin(revenue))


# Total annual revenue
print(np.sum(revenue))

# Difference between highest and lowest revenue
print(np.max(revenue)-np.argmin(revenue))
```

```python
#  show the quaterly total revenue
quarter1 = revenue[:3]
quarter2 = revenue[3:6]
quarter3 = revenue[6:9]
quarter4 = revenue[9:]
quarter = []
quarter.append(quarter1)
quarter.append(quarter2)
quarter.append(quarter3)
quarter.append(quarter4)

print(quarter)

```

```python
# The following array contains sales made by 4 salespeople across 3 months.

# sales = np.array([
# [120, 150, 180],
# [200, 170,160],
# [100, 140, 130],
# [180, 190, 210]
# ])

# Rows represent salespeople and columns represent months.

# Find:

# Total sales by each salesperson
# Average sales by each salesperson
# Total sales for each month
# Average sales for each month
# Highest sales achieved by each salesperson
```

```python
sales = np.array([
[120, 150, 180],
[200, 170,160],
[100, 140, 130],
[180, 190, 210]
])
```

```python
# Total sales by each salesperson
print(np.sum(sales, axis= 1))

# Average sales by each salesperson
print(np.mean(sales, axis= 1))

# Total sales for each month
print(np.sum(sales, axis =0))

# Average sales for each month
print(np.mean(sales, axis= 0))

# Highest sales achieved by each salesperson
print(np.max(sales,axis=1))
```

```python
# Consider:

# employees = np. array([
# [101, 35000, 2],
# [102, 48000, 4],
# [103, 62000, 6],
# [104,40000, 3],
# [105, 75000,8],
# [106, 55000, 5]
# ])
# Columns represent:

# Employee ID | Salary | experience

# Find:

# Employees earning more than ₹50,000.
# Employees with more than 4 years of experience.
# Employees earning more than ₹50,000 AND having more than 4 years of experience.
# Average salary of employees satisfying condition 3.
# Employee with the highest salary.
# Employees earning below the median salary.

```

```python
employees = np. array([
[101, 35000, 2],
[102, 48000, 4],
[103, 62000, 6],
[104,40000, 3],
[105, 75000,8],
[106, 55000, 5]
])
```

```python
# Employees earning more than ₹50,000.
print(employees[:,1][employees[:,1]> 50000])

# Employees with more than 4 years of experience.
print(employees[:,2][employees[:,2]> 4])


# Employees earning more than ₹50,000 AND having more than 4 years of experien
print(employees[(employees[:,1] >50000) & (employees[:,2]> 4)])

# Average salary of employees satisfying condition 3.
print(np.average(employees[:,1]))

# Employee with the highest salary.
print(np.max(employees[:,1]))

# Employees earning below the median salary.
employees[:,1] > np.median(employees[:,1])
print(employees[employees[:,1] > np.median(employees[:,1])])
```
