```python
# Convert full names into lowercase usernames by joining the first and last names with a dot.
names = [
"Aman Sharma",
"Riya Verma",
"Karan Patel"
]
names2=[]


for name in names:
  a = name.lower()
  for n in a:
    if(n!=' '):
      names2.append(n)

print(names2)
```
