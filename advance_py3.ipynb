```python
# Generate all possible combinations of T-shirt colors and sizes

colors = ["white","black","blue","green","orange"]
size = [12,13,14,15,16]
```

```python
combinations = [(color,s) for color in colors for s in size]
combinations
```

```python
# Generate pairs whose sum is greater than 12

numbers1 = [1, 2, 3]
numbers2 = [10,20]


pairs = [(n1,n2)  for n1 in numbers1 for n2 in numbers2  if  n1 + n2 > 12]
```

```python
pairs
```

```python
# Find pairs of numbers from 1 to 5 where both numbers are different and their product is even
```

```python
numbers1 = [1,2,3,4,5]
numbers2 = [1,2,3,4,5]
```

```python
fulfilled = [(i,j) for i in numbers1 for j in numbers2 if i != j and (i * j )%2 == 0]
```

```python
fulfilled
```

```python
marks = [
[75,80,90],
[60,85,70],
[95,88,92]
]
```

```python
flatten= []
for mark in marks:
  for m in mark:
    flatten.append(m)
```

```python
flatten
```

```python
flatten = [m for mark in marks for m in mark]
flatten
```

```python
# Extract only passing marks from every row
# marks >= 75

marks = [
[75,80,90],
[60,85,70],
[95,88,92]
]
```

```python
passing_mark = [m for mark in marks for m in mark if m >= 75]
passing_mark
```

```python
# Increase every number by 5

marks = [
[75,80,90],
[60,85,70],
[95,88,92]
]

bonusplusmark = [m+5 for mark in marks for m in mark ]
print(bonusplusmark)
```

```python
# Replace negative values with zero in a matrix

datas = [
[10,-5,20],
[-3, 15,-8],
[25,0,12]
]
```

```python
replacedlist = [d if d > 0 else 0 for data in datas for d in data ]
replacedlist
```
