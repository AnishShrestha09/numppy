```python
# The coordinates of eight delivery locations are:
import numpy as np
points = np.array([
[3,4],
[5,12],
[8,15],
[7,24],
[9,40],
[12,35],
[15,20],
[6,8]
])
# Each row contains:

# [x, y]

# Calculate distance from the origin using:

# distance = V(x2 +y2)

# I
```

```python
# Extract x and y coordinates.
x = points[:,0] # x coordinate
y = points[:,1] # y coordinate

# Calculate distance for every location.
distance = np.sqrt(x**2 + y**2)
print(distance)

# Round distances to 2 decimals.
print(np.round(distance,2))

# Find nearest and farthest location.
print(np.max(distance))
print(np.min(distance))

# Find average distance.
print(np.mean(distance))

# Find locations above average distance.
criteria = distance > np.mean(distance)
print(points[criteria])

# Categorize:
# if >= 30 + Long
# elseif >= 15 - Medium
print(np.where(distance> 30,"long",np.where(distance > 15,"medium","short")))
```

```python
# Find percentage of Long-distance deliveries.

a = np.where(distance> 30,"long",np.where(distance > 15,"medium","short"))
b = len(a[a=="long"])
c = len(a)

d = b/c * 100
print(d)

# Flatten the coordinate array.
print(points.flatten())

# Find the size of the flattened array.
print(points.size)
```
