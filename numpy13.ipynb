```python
# The coordinates of eight delivery locations are:
import numpy as np
points = np.array([
[3,4],
[5,12],
[8,15],
[7,24],
[9,40],
[12,35],
[15,20],
[6,8]
])
# Each row contains:

# [x, y]

# Calculate distance from the origin using:

# distance = V(x2 +y2)

# I
```

```python
# Extract x and y coordinates.
x = points[:,0] # x coordinate
y = points[:,1] # y coordinate

# Calculate distance for every location.
distance = np.sqrt(x**2 + y**2)
print(distance)

# Round distances to 2 decimals.
print(np.round(distance,2))

# Find nearest and farthest location.
print(np.max(distance))
print(np.min(distance))

# Find average distance.
print(np.mean(distance))

# Find locations above average distance.
criteria = distance > np.mean(distance)
print(points[criteria])

# Categorize:
# if >= 30 + Long
# elseif >= 15 - Medium
print(np.where(distance> 30,"long",np.where(distance > 15,"medium","short")))
```

```python
# Find percentage of Long-distance deliveries.

a = np.where(distance> 30,"long",np.where(distance > 15,"medium","short"))
b = len(a[a=="long"])
c = len(a)

d = b/c * 100
print(d)

# Flatten the coordinate array.
print(points.flatten())

# Find the size of the flattened array.
print(points.size)
```

```python
# Consider the following e-commerce dataset:
import numpy as np
orders = np.array([
[101, 3, 1200, 4.5],
[102, 5, 800, 3.8],
[103, 2, 2500,4.8],
[104, 8, 450, 4.1],
[105, 4, 1800, 3.5],
[106, 6, 950, 4.7],
[107, 3,2200, 4.9],
[108, 10, 600, 3.2],
[109, 7, 1500, 4.6],
[110, 2, 3000, 4.0],
[111, 5, 750, 3.9],
[112, 9, 1250, 4.8]
])
# Order ID # Quantity # Price # Rating
```

```python
# Inspect dimensions, shape and size.
print(orders.ndim)
print(orders.shape)
print(orders.size)

# Extract the first 5 orders.
print(orders[:5])

# Extract the last 3 orders.
print(orders[-3:])

# Extract every second order.
print(orders[::2])

# Calculate revenue for every order.
quantity = orders[:,1]
price = orders[:,2]
revenue = price * quantity
print(revenue)

# Calculate total and average revenue.
print(np.sum(revenue))
print(np.average(revenue))

# Find the order with highest revenue.
print(orders[np.argmax(revenue)])

# Calculate median revenue.
print(np.median(revenue))

# Categorize orders:
# Revenue >= 15,000 + High
# Revenue >= 7,500 + Medium
# Otherwise + Low

categorize = np.where(revenue > 15000,"high",np.where(revenue > 7500,"medium","low"))
print(categorize)

# Find orders where quantity >= 5 AND rating >= 4.5.
cat = ((orders[:,1] >=5) &(orders[:,3] >=4.5) )
print(orders[cat])


# Find orders where price is above median OR rating is below 4.
median = np.median(orders[:,2])
cate = ((orders[:,2]> median) & (orders[:,3] <4.5))
orders[cate]


# Give 10% discount to products whose price is above ₹1,000.
discount_price = np.where(orders[:,2]>1000,0.1*orders[:,2],0)
print(discount_price)

# Round discounted prices to 2 decimals.
print(np.round(discount_price,2))

# Calculate standard deviation of price.
print(np.std(orders[:,2]))

print("Find 25th and 75th percentile of price.")
print(np.percentile(orders[:,2],25))
print(np.percentile(orders[:,2],75))


# Find prices above the 75th percentile.
print('# Find prices above the 75th percentile.')
print(orders[orders[:,2]>np.percentile(orders[:,2],75)])

# Find percentage of orders with rating > 4.5.
print("# Find percentage of orders with rating > 4.5.")
totalrating = len(orders[orders[:,-1]>4.5])
total = len(orders[:,-1])
percentage = totalrating /total * 100
print(percentage)


# Reshape revenue into 3 x 4.
reshaped = revenue.reshape(3,4)
print(reshaped)

# Calculate average revenue of each row.
print(np.average(reshaped, axis = 1))

# Flatten the revenue array.
print(reshaped.flatten())

# Calculate total revenue from orders having rating >= 4.5 AND revenue above average revenue.
categori = ((orders[:,-1]>4.5) & (revenue > np.average(revenue)))
print(np.sum(revenue[categori]))
```
