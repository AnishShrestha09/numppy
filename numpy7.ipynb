```python
# A company stores quarterly revenue and expenses:

# financials = np.array([
# [120, 80],
# [150,90],
# [180, 110],
# [200, 130]
# ])
# Rows = Q1, 02, 03, 04
# Columns = Revenue, Expense

# Tasks
# Display Q2's complete financial data.
# Display all quarterly revenues.
# Display all quarterly expenses.
# Extract Q2 and Q3 data.
# Increase Q4 revenue by 15%.
# Reduce Q1 expense by 10%.
# Display the final financial matrix.
```

```python
financials = np.array([
[120, 80],
[150,90],
[180, 110],
[200, 130]
])
```

```python
# Display Q2's complete financial data.
financials[1]
```

```python
# Display all quarterly revenues.
financials[:,0]
```

```python
# Display all quarterly expenses.
financials[:,1]
```

```python
# Extract Q2 and Q3 data.
financials[1:3]
```

```python
# Increase Q4 revenue by 15%.
financials[-1, 0] = 1.15 * financials[-1, 0]
```

```python
# Reduce Q1 expense by 10%.
financials[0,0] =0.9 * financials[0,0]
```

```python
# Display the final financial matrix.
financials
```

```python
# A classroom has 6 rows and 5 seats per row.

# Initially, all seats are empty.

# 0 = Empty
# 1 = Occupied
# Tasks
# Create a 6
# Mark the first seat of the first row as occupied.
# Mark all seats in the 3rd row as occupied.
# Mark the last two seats of the last row as occupied.
# Display the first 3 rows.
# Display the last 2 columns.
# Display the complete seating matrix.

# 5 array filled with zeros.
```

```python
classroom = np.full((6,5),0)
```

```python
# Mark the first seat of the first row as occupied.
classroom[0,0] = 1
```

```python
classroom
```

```python
# Mark all seats in the 3rd row as occupied.
classroom[2,:] = 1
classroom
```

```python
# Mark the last two seats of the last row as occupied.
classroom[-1,-2:] = 1
print(classroom)
```

```python
# Display the first 3 rows.
classroom[0:3]
```

```python
# Display the last 2 columns.
classroom[:,-2:]
```

```python
# Display the complete seating matrix.

classroom
```
