```python
#

# even_odd = [."Even", "Odd"]

# [value_if_true if condition else value if false for item in iterable]

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_odd = ["even" if n%2 == 0 else "odd" for n in numbers]

print(even_odd)
```

```python
# conditional expressions when there are three categories.

# Marks ≥ 80 "Distinction"

# Marks ≥ 40 - "Pass"

# Otherwise "Fail"


# We can use nested
```

```python
# Problem statement: Given employee salaries, calculate bonuses using these rules:

# Salary ≥ <80,000: 20% bonus

# Salary ≥ <50,000 but below <80,000: 10% bonus

# Salary below ₹50,000: 5% bonus

# Return a list containing the bonus amounts, not the updated salaries.

```

```python
salaries = [90000, 65000, 40000, 100000, 50000]

bonus_amount = [0.2 * salary if salary > 80000
              else 0.1 *salary if salary >=50000
              else 0.05 *salary
              for salary in salaries]

print(bonus_amount)
```

```python
# Classify Online Orders by Value

# Problem statement: Given a list of order values, assign shipping categories using these rules:

# Order value ≥ ₹2,000: "Free Express Shipping"

# Order value ≥ ₹1,000 but below ₹2,000: "Free Standard Shipping"

# Order value below ₹1,000: "Shipping Fee Applicable"
```

```python
orders = [2500, 1500, 800, 3000, 1200, 500]

classify = ["Free Express Shipping" if order > 2000
            else "freestandardshipping" if order >=1000 and order <=2000
            else "shipping fee"
            for order in orders]
print(classify)
```

```python
# Given two lists, find elements that are present in the first list but not in the second list.

List1= [10, 20, 30, 40]
List2= [20, 40, 50]
List3 = []

for elem in List1:
  if elem  not in List2:
    List3.append(elem)

print(List3)
```

```python
# Given a list of integers, determine whether the list is sorted in ascending order.

# Do not use sort() or sorted()-
```

```python
# Input: [10, 20, 30, 40]
# Output: Sorted

# Input: [10, 30, 20, 40]
# Output: Not Sorted
```

```python
Input= [30, 20, 30, 40]
num = Input[0]
for i in range(1,len(Input)):
  if num < Input[i]:
    flag = True
    num = Input[i]
  else:
    flag = False
    break

print(flag)
```
