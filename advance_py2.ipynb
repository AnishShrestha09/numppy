```python
# list comprehension
```

```python
numbers = [1,2,3,4,5]
```

```python
square = []

for i in numbers:
  square.append(i ** 2)
print(square)
```

```python
numbers = [1,2,3,4,5]
square = [element**2 for element in numbers]
print(square)
```

```python
# 5 times multiples of a list
```

```python
numbers = [1, 4, 9, 16, 25]

fivetimes = []

for num in numbers:
  fivetimes.append(num*5)

print(fivetimes)
```

```python
numbers = [1, 4, 9, 16, 25]
fivetimes = [n * 5 for n in numbers]
print(fivetimes)
```

```python
# Suppose a teacher wants to add 5 bonus marks to every student.

numbers = [1, 2, 3, 4, 5]
bonusplusnum = [n + 5 for n in numbers]
print(bonusplusnum)
```

```python
bonus = []
for n in numbers:
  bonus.append(n+5)

print(bonus)
```

```python
# Convert Celsius temperatures to Fahrenheit using:
# F=(9/5)*C+32

celsius = [0, 10, 20, 30, 40]
farehnit =[]
for i in celsius:
  farehnit.append((9/5)*i+32)

print(farehnit)

```

```python
celsius = [0, 10, 20, 30, 40]
farehnit = [(9/5)*c+32 for c in celsius]
```

```python
farehnit
```

```python
# Convert Names to Uppercase

names = ["amit", "neha", "rahul", "priya"]
capital = []
for n in names:
  capital.append(n.upper())
print(capital)
```

```python
names = ["amit", "neha", "rahul", "priya"]
capital = [n.upper() for n in names]
```

```python
capital
```

```python
prices = [100.50, 250.75, 399.99, 120.25]
# convert each price to floor value integer.
floorprice = [int(price) for price in prices]
print(floorprice)

```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even = []

for n in numbers:
  if n % 2 == 0:
    even.append(n)

print(even)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even = [n for n in numbers  if n % 2 == 0]
print(even)
```

```python
# Find Numbers Greater Than 50

numbers = [15, 60, 45, 90, 35, 75]
greaterthan50 = [n for n in numbers if n > 50 ]
print(greaterthan50)
```

```python
words = ["cat", "elephant", "dog", "giraffe", "ant"]
# create a list of only those words which are of atleast length 5
```
