```python
# A bank account has the following transactions:
import numpy as np

transactions = np.array([
5000, -2000, 15000, -4500, 7000,
-12000, 25000, -3000, 8000, -18000,
4500, 22008, -6008, 9000, -2500
])

# Extract deposits.
print(transactions[transactions > 0 ])


# Extract withdrawals.
print(transactions[transactions < 0 ])


# Calculate total deposits and withdrawals.
print(np.sum(transactions[transactions > 0 ]))
print(np.sum(transactions[transactions < 0 ]))
sum_pos = np.sum(transactions[transactions > 0 ])
sumneg = np.absolute(np.sum(transactions[transactions < 0 ]))


# Calculate absolute transaction values.
total_transaction = sum_pos + sumneg
print(total_transaction)


# Calculate net transaction amount.
a = np.sum(transactions[transactions > 0 ])-np.sum(transactions[transactions < 0 ])
print(a)
```

```python
# Find transactions whose absolute value is above the average absolute transaction.
mean = np.mean(np.absolute(transactions))
print(transactions[np.absolute(transactions) > mean])

# Find the largest deposit.
print(np.max(transactions))

# Find the largest withdrawal.
print(np.min(transactions))

# Create labels "Credit" and "Debit".
np.where(transactions > 0 ,"credit","debit")

# Find pergentage of transactions whose absolute value is greater than ₹10,000.
absolute_transaction = np.absolute(transactions)
a = len(transactions[transactions > 10000])
total = len(transactions)
percentage = a/total * 100
print(percentage)

# Reshape the data into 3 X 5.
b = transactions.reshape(3,5)
print(b)

# Calculate total transaction value for each row.
absolute_b = np.absolute(b)
print(np.sum(absolute_b, axis = 1))



```

```python
# A factory has four machines operating for six days:
import numpy as np

production = np.array([
[120, 135, 128, 145, 150, 138],
[110, 125, 118, 130, 142, 135],
[160, 155, 170, 165, 180, 175],
[95, 105, 100, 115, 108, 120]
])
```

```python
# Calculate total production by machine.
print(np.sum(production ,axis = 1))

# Calculate total production by day.
print(np.sum(production, axis = 0))

# Calculate average production of each machine.
average = np.average(production, axis =1 )
print(average)

# Find the machine with the highest average production.
ind = np.argmax(average)
print(production[ind])


# Find the day with the lowest production.
abs = np.sum(production, axis = 0)
a = np.argmin(abs)
print(production[a])
```

```python
# Consider the following loan portfolio:

loan_amount = np.array([
100000, 250000, 500000, 750000,
120000, 350000, 900000, 450000
])

interest_rate = np.array([
8.5, 9.2, 8.8, 10.1,
7.9, 9.5, 10.5, 8.7
])
```

```python
# Calculate annual interest for every loan.
annual_interest = (loan_amount * interest_rate)/100
print(annual_interest)

# Calculate total interest.
total_interest = np.sum(annual_interest)
print(total_interest)

# Calculate average interest.
print(np.mean(annual_interest))

# Find the loan generating the highest interest.
print(np.max(annual_interest))

# Find loans above ₹400,000 with interest rate above 9%.
interest_rate1 = (interest_rate> 9) & (loan_amount > 400000)
print(loan_amount[interest_rate1])
```
