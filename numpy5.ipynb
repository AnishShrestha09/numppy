```python
import numpy as np
```

```python
marks = np.array([70,80,65,90,75])
```

```python
marks[1:3] # last index is excluded
```

```python
marks[2:4]
```

```python
marks[0:4:2] #start,end,step
```

```python
marks[:3] # starting is zero
```

```python
marks[3:] #end is last index
```

```python
marks[::] # start end step is default means 1
```

```python
marks[::2] # print all the alternate elements
```

```python
numbers = np.array([10, 20, 30, 40, 50, 60, 70])

print(numbers[1:4]) # 20 30 40
print(numbers[2:6]) # 30 40 50 60
print(numbers[0:3]) # 10 20 30
```

```python
numbers = np.array([10,20, 30, 40, 50])

print(numbers[2:]) # 30 40 50
```

```python
sales = np.array([100, 150, 200, 250, 300, 350])

print(sales[-3:]) # 250 300 350
```

```python
sales = np.array([100, 150, 200, 250, 300, 350])

print(sales[ :- 2]) # 100, 150, 200, 250
```

```python
numbers = np.array([10,20,30,40, 50, 60])

print(numbers [ :: 3]) #10 40
```

```python
# reverse the given array
numbers[::-1] # -1 is the step always come from backward
```
